```python
# step 1 : Import Required libaraies
import pandas as pd
import numpy as np
```

```python
#step 2 : load dataset
df = pd.read_excel("Ecommerce_Unclean_Project.xlsx")
```

```python
#step 4: check missing value
df.isnull().sum()
```

```python
#step 5: check duplicate
df.duplicated().sum()
```

```python
#remove duplicate
df.drop_duplicates(inplace=True)
df.duplicated().sum()
```

```python
#step 6: replace invalid values
# replace common inc=valid values with NaN
df.replace(['N/A','NULL',''],np.nan,inplace=True)
df
```

```python
#step 7: Remove extra spaces
df= df.apply(lambda x: x.str.strip() if x.dtype =='object' else x)
```

```python
# step 8: standardize text format
# convert text into proper case

df['Customer_Name'] = df['Customer_Name'].str.title()
df['City'] = df['City'].str.title()
df['State'] = df['State'].str.title()
```

```python
# step 9: Validate Email Address
# keep only valid email record

df = df[df['Email'].str.contains('@',na=False)]
```

```python
#step 10: covert date column
df['Order_Date']=pd.to_datetime(df['Order_Date'],errors='coerce')
df['Delivery_Date']=pd.to_datetime(df['Delivery_Date'],errors='coerce')
```

```python
# step 11 : Convert Numeric columns
cols = ['Qty','Unit_Price','Discount']
for c in cols:
    df[c] = pd.to_numeric(df[c],errors='coerce')
```

```python
# step 12: Remove invalid quantity
# remove records having negative or zero quantity

df = df[df['Qty']>0]
```

```python
# step 13: fill missing value
df['Discount']= df['Discount'].fillna(0)
df['Phone'] = df['Phone'].fillna('Unknow')
df['Delivery_Date'] = df['Delivery_Date'].fillna(df['Order_Date'])
```

```python
# step 14: Feature engineering
#total sales
df['Sales']= df['Qty']*df['Unit_Price']
```

```python
#net amt
df['Net_amount']= df['Sales']-(df['Sales']*df['Discount']/100)
```

```python
df
```

```python
#profit 
df['Profit']= df['Net_amount']*0.20
```

```python
df
```

```python
#Order month
df['Month'] = df['Order_Date'].dt.month_name()
```

```python
df
```

```python
# order year
df['Year'] = df['Order_Date'].dt.year
```

```python
df
```

```python
# week year
df['Weekday'] = df['Order_Date'].dt.day_name()
```

```python
df
```

```python
#step 15 : final data
df.info()
```
